```python
import requests
import json
import csv
from pathlib import Path
```

```python
API_URL = "https://world.openfoodfacts.org/api/v2/search"

HEADERS = {
    "User-Agent": "RowadMasr-Python-MohamedFarag/1.0 (themohamedfarag@gmail.com)"
}

PARAMS = {
    "categories_tags_en": "breakfast-cereals",
    "page_size": 100,
    "fields": "code,product_name,brands,quantity,categories_tags_en,countries_tags,ingredients_text,nutrition_grades,nutriments",
}
```

```python
response = requests.get(API_URL, params=PARAMS, headers=HEADERS)
payload = response.json()
all_records = payload["products"]

print(f"Fetched {len(all_records)} products")
```

```python
missing_key = 0
empty_string = 0
has_value = 0

for record in all_records:
    if "ingredients_text" not in record:
        missing_key += 1
    elif record.get("ingredients_text") == "":
        empty_string += 1
    else:
        has_value += 1

print(f"Missing key entirely: {missing_key}")
print(f"Present but empty string: {empty_string}")
print(f"Has real value: {has_value}")
```

```python
missing_grade = 0
has_grade = 0

for record in all_records:
    if "nutrition_grades" not in record:
        missing_grade += 1
    else:
        has_grade += 1

print(f"Missing nutrition_grades key: {missing_grade}")
print(f"Has nutrition_grades key: {has_grade}")
```

```python
nutrient_fields = ["sugars_100g", "fat_100g", "fiber_100g", "salt_100g", "proteins_100g", "energy-kcal_100g"]

for field in nutrient_fields:
    count = 0
    for record in all_records:
        if record.get("nutriments", {}).get(field) is not None:
            count += 1
    pct = (count / len(all_records)) * 100
    print(f"{field}: {count}/{len(all_records)} ({pct:.1f}%)")
```

```python
total = 0
count = 0
min_val = None
max_val = None
missing = 0

for record in all_records:
    value = record.get("nutriments", {}).get("sugars_100g")
    if value is None:
        missing += 1
        continue
    total += value
    count += 1
    if min_val is None or value < min_val:
        min_val = value
    if max_val is None or value > max_val:
        max_val = value

mean_val = total / count if count else None

print(f"Missing: {missing}")
print(f"Count used: {count}")
print(f"Min: {min_val}")
print(f"Max: {max_val}")
print(f"Mean: {mean_val:.2f}")
```

```python
def explore_structure(data, indent=0):
    prefix = "  " * indent
    if isinstance(data, dict):
        for key, value in data.items():
            print(f"{prefix}{key}: {type(value).__name__}")
            explore_structure(value, indent + 1)
    elif isinstance(data, list):
        print(f"{prefix}[list of {len(data)} items]")
        if data:
            explore_structure(data[0], indent + 1)
    else:
        pass  
```
